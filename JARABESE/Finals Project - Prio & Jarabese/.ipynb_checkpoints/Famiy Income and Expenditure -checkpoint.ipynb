```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('datasets\\Family Income and Expenditure.csv')
```

```python
df
```

<hr>
<h2>1. Average Household Income

 - Chart type: Histogram or Boxplot<br>
 - Purpose: Show income distribution, highlight mean/median.</h2>

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Total Household Income'], bins=30, kde=True, color='skyblue')
plt.axvline(df['Total Household Income'].mean(), color='red', linestyle='--', label='Mean')
plt.axvline(df['Total Household Income'].median(), color='blue', linestyle='--', label='Median')
plt.title("Distribution of Household Income")
plt.legend()
plt.show()
```

<hr>
<h2>2. Income by Region
    
 - Chart type: Bar chart <br>

 - Purpose: Compare average income across regions.</h2>

```python
region_income = df.groupby('Region')['Total Household Income'].mean().sort_values()
plt.figure(figsize=(10,6))
sns.barplot(x=region_income.index, y=region_income.values, palette="viridis")
plt.xticks(rotation=55
plt.title("Average Household Income by Region")
plt.ylabel("Average Income")
plt.show()
```

<hr>
<h2>3. Main Source of Income

 - Chart type: Pie chart <br>

 - Purpose: Show proportion of households by income source.</h2>

```python
plt.figure(figsize=(6,6))
df['Main Source of Income'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Main Source of Household Income")
plt.ylabel("")
plt.show()
```

<hr>
<h2>4. Household Size & Education Profile

 - Chart type: Two visuals

   - Histogram for household size

   - Bar chart for education levels</h2>

```python
size_col = [col for col in df.columns if 'size' in col.lower() or 'member' in col.lower()]
if size_col:
    plt.figure(figsize=(8,5))
    sns.histplot(df[size_col[0]], bins=10, color='green')
    plt.title("Distribution of Household Size")
    plt.show()
else:
    print("⚠️ Household size column not found.")
```

<hr>
<h2>5. Basic Spending (Food, Utilities, etc.)

 - Chart type: Stacked bar chart or Pie chart

 - Purpose: Show spending composition.</h2>

```python
# Define expected spending columns
spending_cols = [
    'Total Food Expenditure',
    'Bread and Cereals Expenditure',
    'Total Rice Expenditure',
    'Meat Expenditure',
    'Total Fish and marine products Expenditure',
    'Fruit Expenditure',
    'Utilities Expenditure'  # Add if available
]

# Filter only columns that exist in the dataset
valid_spending_cols = [col for col in spending_cols if col in df.columns]

# Compute average spending
spending_share = df[valid_spending_cols].mean()

# Plot pie chart
plt.figure(figsize=(7,7))
spending_share.plot(kind='pie', autopct='%1.1f%%')
plt.title("Average Household Spending Breakdown")
plt.ylabel("")
plt.show()
```
